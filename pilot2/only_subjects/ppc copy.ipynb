```python
def calculate_phase_at_frequencies(signal_data, fs, freq_bands):
    """
    Calculate phase at different frequency bands using Hilbert transform
    
    Parameters:
    -----------
    signal_data : array
        Input time series data
    fs : float
        Sampling frequency in Hz
    freq_bands : list of tuples
        List of (low_freq, high_freq) defining frequency bands of interest
        
    Returns:
    --------
    phases : dict
        Dictionary with frequency bands as keys and phase arrays as values
    """
    phases = {}
    
    for low_freq, high_freq in freq_bands:
        # Design a bandpass filter
        nyquist = fs / 2
        low = low_freq / nyquist
        high = high_freq / nyquist
        b, a = signal.butter(2, [low_freq, high_freq], fs = fs, btype='bandpass')
        
        # Apply the filter
        filtered_signal = signal.filtfilt(b, a, x=signal_data)
        
        # Apply Hilbert transform to get analytic signal
        analytic_signal = hilbert(filtered_signal)
        
        # Extract instantaneous phase
        instantaneous_phase = np.angle(analytic_signal)
        
        # Store in dictionary
        band_name = f"{low_freq}-{high_freq}Hz"
        phases[band_name] = instantaneous_phase
    
    return phases

```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import circmean


events    = ['familiar', 'novel', 'cagemate']
freq_bands = ['4-12Hz', '12-30Hz', '30-59Hz', '59-80Hz']

records = []  # single list, outside all loops

for region in only_subjects_lfp.brain_region_dict.keys():
    for recording in only_subjects_spike.recordings:
        for unit, spike_trains in recording.unit_spiketrains.items():
            
            # --- precompute all phases for this recording × region ---
            # shape: {freq_band: array of shape (T,)}
            phase_arrays = {
                fb: rec_phase_dict[recording.name][region][fb]
                for fb in freq_bands
            }
            
            for event in events:
                times = epoch_map_dict[recording.name][event]

                # slice spike train once
                try:
                    t0 = int(times[0]) * 1000
                    t1 = int(times[1]) * 1000
                    seg_spike_train = spike_trains[t0:t1]
                except (IndexError, ValueError, TypeError):
                    t0 = int(times[0]) * 1000
                    t1 = int(spike_trains.shape[0])
                    seg_spike_train = spike_trains[t0:t1]

                # now do all freq bands with the same spike slice
                for freq_band in freq_bands:
                    seg_lfp_phase = phase_arrays[freq_band][t0:t1]

                    # align lengths
                    if seg_spike_train.shape[0] != seg_lfp_phase.shape[0]:
                        shorter = min(seg_spike_train.shape[0], seg_lfp_phase.shape[0])
                        seg_spike_train_fb = seg_spike_train[:shorter]
                        seg_lfp_phase      = seg_lfp_phase[:shorter]
                    else:
                        seg_spike_train_fb = seg_spike_train

                    n_spikes = int(sum(seg_spike_train_fb))

                    if n_spikes > 100:
                        mask         = seg_spike_train_fb.astype(bool)
                        spike_phases = seg_lfp_phase[mask]
                        n            = len(spike_phases)
                        z            = np.exp(1j * spike_phases)
                        R            = float(np.abs(np.sum(z))**2)
                        PPC          = (R - n) / (n * (n - 1))
                        p            = rayleightest(spike_phases)
                        pref_phase   = circmean(spike_phases, high=np.pi, low=-np.pi)
                    else:
                        PPC        = np.nan
                        p          = np.nan
                        pref_phase = np.nan
                        n_spikes   = np.nan

                    records.append({
                        'unit'      : f'{recording.name}_{unit}',
                        'recording' : recording.name,
                        'region'    : region,
                        'event'     : event,
                        'freq_band' : freq_band,
                        'PPC'       : PPC,
                        'rayleigh_p': p,
                        'pref_phase': pref_phase,
                        'n_spikes'  : n_spikes,
                    })

df = pd.DataFrame(records)
for region in df['region'].unique():
    # ── Plot 1: PPC per cell across conditions (one subplot per freq band) ──
    fig, axes = plt.subplots(1, len(freq_bands), figsize=(16, 5), sharey=False)
    fig.suptitle(f'{region} — PPC per cell across conditions', fontsize=13)
    condition_colors = {'familiar': '#3498db', 'novel': '#e74c3c', 'cagemate': '#2ecc71'}

    for ax, freq_band in zip(axes, freq_bands):
        # pivot so each row is a unit, columns are conditions
        pivot = (df[df.freq_band == freq_band]
                   .pivot_table(index='unit', columns='event', values='PPC'))
        # only keep units with values in all 3 conditions
        pivot = pivot.dropna()

        for i, (cond, color) in enumerate(condition_colors.items()):
            if cond not in pivot.columns:
                continue
            x = np.full(len(pivot), i) + np.random.uniform(-0.08, 0.08, len(pivot))
            ax.scatter(x, pivot[cond], alpha=0.4, s=18, color=color)
            ax.plot([i - 0.15, i + 0.15],
                    [pivot[cond].median(), pivot[cond].median()],
                    color=color, linewidth=2.5)

        # draw lines connecting same cell across conditions
        for _, row in pivot.iterrows():
            vals = [row.get(c, np.nan) for c in events]
            if not any(np.isnan(vals)):
                ax.plot(range(len(events)), vals,
                        color='gray', alpha=0.15, linewidth=0.7, zorder=0)

        ax.set_title(freq_band, fontsize=10)
        ax.set_xticks(range(len(events)))
        ax.set_xticklabels(events, rotation=20, fontsize=9)
        ax.set_ylabel('PPC' if ax == axes[0] else '')
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.axhline(0, color='black', linewidth=0.5, linestyle='--', alpha=0.4)

    plt.tight_layout()
    plt.show()

    # ── Plot 2: Phase preference polar plots (one per freq band × condition) ──
    fig, axes = plt.subplots(len(events), len(freq_bands),
                             figsize=(14, 10),
                             subplot_kw=dict(projection='polar'))
    fig.suptitle(f'{region} — preferred phase per cell', fontsize=13)

    for row_i, event in enumerate(events):
        for col_i, freq_band in enumerate(freq_bands):
            ax = axes[row_i][col_i]
            sub = df[(df.event == freq_band) & (df.freq_band == freq_band)].copy()
            sub = df[(df.event == event) & (df.freq_band == freq_band)].copy()

            # only significant cells
            sig = sub[sub.rayleigh_p < 0.05]['pref_phase'].dropna()
            all_cells = sub['pref_phase'].dropna()

            # all cells in gray
            ax.scatter(all_cells, np.ones(len(all_cells)) * 0.6,
                       alpha=0.3, s=12, color='gray')
            # significant cells colored
            if len(sig) > 0:
                ax.scatter(sig, np.ones(len(sig)) * 0.6,
                           alpha=0.7, s=18,
                           color=condition_colors[event])
                # mean resultant vector for sig cells
                mean_phase = circmean(sig, high=np.pi, low=-np.pi)
                ax.annotate('', xy=(mean_phase, 0.5), xytext=(0, 0),
                            arrowprops=dict(arrowstyle='->', color=condition_colors[event],
                                            lw=2))

            ax.set_yticklabels([])
            ax.set_theta_zero_location('N')
            if row_i == 0:
                ax.set_title(freq_band, fontsize=9)
            if col_i == 0:
                ax.set_ylabel(event, fontsize=9, labelpad=30)

    plt.tight_layout()
    plt.show()
```

```python
condition_colors = {'novel': '#15616F', 'familiar': '#FFAF00', 'cagemate': '#792910'}
events = ['cagemate','novel', 'familiar']

for region in df['region'].unique():
    region_df = df[df['region'] == region].copy()

    # ── Plot 1: PPC per cell across conditions ──
    fig, axes = plt.subplots(1, len(freq_bands), figsize=(16, 5), sharey=False)
    fig.suptitle(f'{region} — PPC per cell across conditions', fontsize=13)

    for ax, freq_band in zip(axes, freq_bands):
        pivot = (region_df[region_df.freq_band == freq_band]
           .pivot_table(index='unit', columns='event', values='PPC')
           .dropna()
           .reindex(columns=events))

        for i, (cond, color) in enumerate(condition_colors.items()):
            if cond not in pivot.columns:
                continue
            x = np.full(len(pivot), i) + np.random.uniform(-0.08, 0.08, len(pivot))
            ax.scatter(x, pivot[cond], alpha=0.4, s=18, color=color)
            ax.plot([i - 0.15, i + 0.15],
                    [pivot[cond].median(), pivot[cond].median()],
                    color=color, linewidth=2.5)

        for _, row in pivot.iterrows():
            vals = [row.get(c, np.nan) for c in events]
            if not any(np.isnan(vals)):
                ax.plot(range(len(events)), vals,
                        color='gray', alpha=0.15, linewidth=0.7, zorder=0)

        ax.set_title(freq_band, fontsize=10)
        ax.set_xticks(range(len(events)))
        ax.set_xticklabels(events, rotation=20, fontsize=9)
        ax.set_ylabel('PPC' if ax == axes[0] else '')
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.axhline(0, color='black', linewidth=0.5, linestyle='--', alpha=0.4)

    plt.tight_layout()
    plt.show()

    # ── Plot 2: Phase preference polar plots ──
    fig, axes = plt.subplots(len(events), len(freq_bands),
                             figsize=(14, 10),
                             subplot_kw=dict(projection='polar'))
    fig.suptitle(f'{region} — preferred phase per cell', fontsize=13)

    for row_i, event in enumerate(events):
        for col_i, freq_band in enumerate(freq_bands):
            ax = axes[row_i][col_i]
            sub = region_df[(region_df.event == event) & 
                            (region_df.freq_band == freq_band)]

            all_cells = sub['pref_phase'].dropna()
            sig       = sub[sub.rayleigh_p < 0.05]['pref_phase'].dropna()

            ax.scatter(all_cells, np.ones(len(all_cells)) * 0.6,
                       alpha=0.3, s=12, color='gray')

            if len(sig) > 0:
                ax.scatter(sig, np.ones(len(sig)) * 0.6,
                           alpha=0.7, s=18, color=condition_colors[event])
                mean_phase = circmean(sig, high=np.pi, low=-np.pi)
                ax.annotate('', xy=(mean_phase, 0.5), xytext=(0, 0),
                            arrowprops=dict(arrowstyle='->',
                                            color=condition_colors[event], lw=2))

            ax.set_yticklabels([])
            ax.set_theta_zero_location('N')
            if row_i == 0:
                ax.set_title(freq_band, fontsize=9)
            if col_i == 0:
                ax.set_ylabel(event, fontsize=9, labelpad=30)

    plt.tight_layout()
    plt.show()
```
